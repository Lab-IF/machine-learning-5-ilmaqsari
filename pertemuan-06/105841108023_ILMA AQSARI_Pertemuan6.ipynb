## Import Library

```python
import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.datasets import load_diabetes

from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import accuracy_score
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score
```

## 1. Load Dataset Classification

```python
print("="*60)
print("LOAD DATASET BREAST CANCER")
print("="*60)

data = load_breast_cancer()

print("Jumlah Data :",data.data.shape)
print("Jumlah Fitur:",len(data.feature_names))
```

```python
print("\nNama Fitur (5 pertama):")
print(data.feature_names[:5])

print("\nContoh Data:")
print(data.data[:5])
```

## 2. Split Data

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

data = load_breast_cancer()

X_train,X_test,y_train,y_test = train_test_split(

    data.data,
    data.target,
    test_size=0.2,
    random_state=42

)

print("\nUkuran Training :",X_train.shape)
print("Ukuran Testing  :",X_test.shape)
```

```python
print("\nDistribusi Kelas:")
print("Kelas 0:",(y_train==0).sum())
print("Kelas 1:",(y_train==1).sum())
```

## 3. Tugas 1 - Perbandingan Model

```python
print("\n" + "="*60)

print("TUGAS 1: PERBANDINGAN MODEL")

print("="*60)


models = {

"Decision Tree":

DecisionTreeClassifier(random_state=42),


"Random Forest":

RandomForestClassifier(
n_estimators=100,
random_state=42
),


"Gradient Boosting":

GradientBoostingClassifier(
random_state=42
)

}


hasil={}


for nama,model in models.items():

    model.fit(X_train,y_train)

    pred=model.predict(X_test)

    acc=accuracy_score(y_test,pred)

    hasil[nama]=acc

    print(nama,"Accuracy =",round(acc,3))
```

```python
print("\nInsight Model:")

print("Decision Tree mudah dipahami tetapi kurang stabil")

print("Random Forest lebih stabil karena menggunakan banyak tree")

print("Gradient Boosting membangun model secara bertahap")
```

## 4. Tugas 2 - Feature Importance

```python
print("\n" + "="*60)

print("TUGAS 2: FEATURE IMPORTANCE")

print("="*60)


rf=models["Random Forest"]

importance=rf.feature_importances_

index=np.argsort(importance)[::-1]


for i in range(10):

    print(
        i+1,
        data.feature_names[index[i]],
        round(importance[index[i]],4)
    )
```

```python
print("\nFeature importance menunjukkan fitur yang paling")
print("berpengaruh dalam menentukan klasifikasi.")
```

## 5. Tugas 3 - Pengaruh n_estimators

```python
print("\n" + "="*60)

print("TUGAS 3: PENGARUH n_estimators")

print("="*60)


for n in [10,50,100,200,500]:

    model=RandomForestClassifier(

        n_estimators=n,

        random_state=42

    )

    model.fit(X_train,y_train)

    pred=model.predict(X_test)

    acc=accuracy_score(y_test,pred)

    print("n_estimators =",n,"Accuracy =",round(acc,3))
```

```python
print("\nInsight n_estimators:")
print("Semakin banyak tree biasanya model lebih stabil")
print("Namun waktu training juga semakin lama")
```

## 6. tambahan - Ensemble Regression

```python
print("\n" + "="*60)

print("ENSEMBLE REGRESSION")

print("="*60)

db=load_diabetes()

X_train_r,X_test_r,y_train_r,y_test_r=train_test_split(

db.data,
db.target,

test_size=0.2,

random_state=42

)

reg=RandomForestRegressor(

n_estimators=100,

random_state=42

)

reg.fit(X_train_r,y_train_r)
pred=reg.predict(X_test_r)


print("RMSE =",round(
root_mean_squared_error(
y_test_r,
pred
),3))


print("R2 =",round(
r2_score(
y_test_r,
pred
),3))
```

```python
print("\nInterpretasi Regression:")
print("RMSE kecil menunjukkan error prediksi rendah")
print("R2 mendekati 1 menunjukkan model cukup baik")
```
