```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, load_wine
```

## 1. NumPy
Operasi dasar array.

```python
x = np.array([1, 2, 3, 4, 5])

print("=== NumPy ===")
print("Array:", x)
print("Mean :", x.mean())
print("Median :", np.median(x))
print("Std :", x.std())
```

## 2. Pandas
Membuat DataFrame sederhana.

```python
df = pd.DataFrame({
    "nama": ["A", "B", "C"],
    "nilai": [80, 75, 90]
})

print("=== Pandas ===")
print(df)

print("\nStatistik nilai:")
print(df["nilai"].describe())
```

## 3. Visualisasi
Bar chart nilai mahasiswa.

```python
df.plot(kind="bar", x="nama", y="nilai", legend=False)

plt.title("Nilai Mahasiswa")
plt.xlabel("Nama")
plt.ylabel("Nilai")
plt.tight_layout()
plt.show()

plt.savefig("pertemuan01_barchart.png")
print("Bar chart disimpan ke pertemuan01_barchart.png")
```

## 4. Dataset Iris

```python
iris = load_iris(as_frame=True)

print("=== Dataset Iris (5 baris pertama) ===")
print(iris.frame.head())

print("\nUkuran dataset:", iris.frame.shape)
print("Jumlah kelas:", iris.frame["target"].nunique())
```

```python
print("=== Tugas 1: Statistik Array Random ===")

np.random.seed(42)
arr = np.random.rand(20) * 100

print("Array:", arr.round(1))
print("Mean :", round(arr.mean(), 2))
print("Median:", round(np.median(arr), 2))
print("Std  :", round(arr.std(), 2))
```

```python
print("=== Tugas 2: Top 3 Nilai ===")

np.random.seed(0)

df_mhs = pd.DataFrame({
    "nama": [f"Mahasiswa_{i+1}" for i in range(10)],
    "nilai": np.random.randint(50, 100, 10)
})

print(df_mhs)

top3 = df_mhs.nlargest(3, "nilai")

print("\nTop 3:")
print(top3)

print("\nRata-rata nilai:", round(df_mhs["nilai"].mean(),2))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(df_mhs["nama"], df_mhs["nilai"])
axes[0].set_title("Bar Chart Nilai")
axes[0].tick_params(axis="x", rotation=45)

axes[1].hist(df_mhs["nilai"], bins=5)
axes[1].set_title("Histogram Nilai")

plt.tight_layout()
plt.show()

plt.savefig("pertemuan01_tugas_grafik.png")
print("Grafik disimpan ke pertemuan01_tugas_grafik.png")
```

```python
wine = load_wine(as_frame=True)

print("=== Tugas 4: Dataset Wine ===")

print("Ukuran data:", wine.frame.shape)
print("Kolom:", list(wine.frame.columns))

print("\nInsight 1:")
print("Dataset Wine memiliki", wine.frame.shape[0], "data dan",
      wine.frame.shape[1]-1, "fitur.")

print("\nInsight 2:")
print("Dataset memiliki", wine.frame["target"].nunique(),
      "kelas wine yaitu", sorted(wine.frame["target"].unique()))

print("\nInsight 3:")
print("Rata-rata alcohol =", round(wine.frame["alcohol"].mean(), 2))
```
