## Import Library

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 1. Load Dataset

```python
print("="*60)
print("LOAD DATASET BREAST CANCER")
print("="*60)

data = load_breast_cancer()

print("Jumlah data :", data.data.shape)
print("Jumlah fitur:", len(data.feature_names))
print("Nama kelas  :", data.target_names)
```

```python
print("\nContoh Data:")
print(data.data[:5])

print("\nContoh Target:")
print(data.target[:5])
```

```python
print("\nStatistik Fitur (5 pertama):")
print(np.round(data.data[:,:5].mean(axis=0),2))
```

## 2. Split Data Training & Testing

```python
X_train, X_test, y_train, y_test = train_test_split(
    data.data,
    data.target,
    test_size=0.2,
    random_state=42
)

print("\nUkuran Training :", X_train.shape)
print("Ukuran Testing  :", X_test.shape)
```

## 3. Scaling Data (WAJIB untuk SVM)

```python
print("\nMelakukan StandardScaler...")

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("\nSetelah scaling:")
print("Mean (mendekati 0):", np.round(X_train.mean(),2))
print("Std (mendekati 1):", np.round(X_train.std(),2))
```

## 4. Tugas 1 - Perbandingan Kernel Linear vs RBF

```python
print("\n" + "="*60)
print("TUGAS 1: PERBANDINGAN KERNEL SVM")
print("="*60)

models = {
    "SVM Linear" : SVC(kernel="linear", C=1),
    "SVM RBF" : SVC(kernel="rbf", C=1, gamma="scale")
}
hasil = {}

for nama, model in models.items():
    model.fit(X_train,y_train)
    pred = model.predict(X_test)
    acc = accuracy_score(y_test,pred)
    hasil[nama] = acc

    print("\nModel :", nama)
    print("Accuracy :", round(acc,3))
    print("\nClassification Report")
    print(classification_report(
        y_test,
        pred,
        target_names=data.target_names
    ))
    print("Confusion Matrix")
    print(confusion_matrix(y_test,pred))
```

```python
print("\nInsight Kernel:")
print("Kernel Linear cocok untuk data yang terpisah secara linear")
print("Kernel RBF lebih fleksibel untuk pola yang kompleks")
```

## 5. Tugas 2 - Tuning Parameter

```python
print("\n" + "="*60)
print("TUGAS 2: TUNING PARAMETER")
print("="*60)

best_accuracy = 0
best_param = {}

for C in [0.1,1,10]:
    for gamma in [0.01,0.1,1]:

        model = SVC(
            kernel="rbf",
            C=C,
            gamma=gamma
        )

        model.fit(X_train,y_train)
        pred = model.predict(X_test)
        acc = accuracy_score(y_test,pred)

        print("C =",C,"gamma =",gamma,"accuracy =",round(acc,3))
        if acc > best_accuracy:
            best_accuracy = acc
            best_param = {
                "C":C,
                "gamma":gamma
            }
print("\nParameter Terbaik")
print(best_param)
print("Best Accuracy:",round(best_accuracy,3))
```

```python
print("\nInterpretasi Parameter:")
print("C besar membuat model lebih fokus pada data training")
print("Gamma besar membuat batas keputusan lebih kompleks")
```

```python
print("\n" + "="*60)
print("KESIMPULAN")
print("="*60)

print("1. Kernel RBF biasanya menghasilkan akurasi lebih baik")
print("   dibanding kernel Linear pada dataset ini.")

print("2. Parameter C dan gamma mempengaruhi batas keputusan SVM.")

print("3. Scaling menggunakan StandardScaler wajib dilakukan")
print("   agar performa SVM optimal.")
```
